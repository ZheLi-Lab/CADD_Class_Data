3.格式转换

（1）SMILES转换为mol：

```python
# MolFromSmiles() 函数读取分子信息。
# AllChem 中的 AddHs() 函数添加氢原子，EmbedMolecule() 函数生成三维坐标，MMFFOptimizeMolecule() 函数进行几何优化。
# MolToMolFile() 函数将分子对象写入 MOL 文件。

import os
from rdkit import Chem
from rdkit.Chem import AllChem

output_path = "output_data"
os.makedirs(output_path, exist_ok=True)

# 定义苯分子的 SMILES
smiles_benzene = 'c1ccccc1'
# 根据 SMILES 字符串创建分子对象, 并展示
mol_benzene = Chem.MolFromSmiles(smiles_benzene)    
display(mol_benzene)
# 向分子中添加氢原子, 并展示
mol_benzene_Hs = AllChem.AddHs(mol_benzene)
display(mol_benzene_Hs)
# 为分子生成三维坐标
AllChem.EmbedMolecule(mol_benzene_Hs)
# 使用MMFF94力场对分子进行几何优化
AllChem.MMFFOptimizeMolecule(mol_benzene_Hs)
# 打印分子对象的类型
print("mol_benzene 对象的类型为：", type(mol_benzene))
print("mol_benzene_Hs 对象的类型为：", type(mol_benzene_Hs))
# 结果文件的保存路径
benzene_path = os.path.join(output_path, "Benzene.mol")
# 将分子对象保存为 mol 文件
Chem.MolToMolFile(mol_benzene_Hs, benzene_path)
print("苯分子的 MOL 文件已保存！路径：", benzene_path)
```

（2）SMILES转换为sdf

```python

# Chem.MolFromSmiles() 函数读取分子信息。
# Draw.MolsToGridImage() 函数可视化分子信息。
# Chem.SDWriter() 函数将分子对象列表写入 SDF 文件。

import os
from rdkit import Chem
from rdkit.Chem import Draw

data_path = "data"
output_path =  "output_data"
os.makedirs(output_path, exist_ok=True)
os.makedirs(data_path, exist_ok=True)
simles_file_path = os.path.join(data_path, "20smiles.smi")
sdf_file_path = os.path.join(output_path, "20smiles_convert.sdf")  

if os.path.exists(simles_file_path):
    print("文件已存在，路径为：", simles_file_path)
else:
    import requests
    file_url = 'https://raw.githubusercontent.com/ZheLi-Lab/CADD_Class_Data/main/chapter6/data/20smiles.smi'
    response = requests.get(file_url)
    with open(simles_file_path, 'wb') as f:
        f.write(response.content)
        print("文件下载成功，保存路径为：", simles_file_path)

# 打开并读取包含 SMILES 字符串的文件
with open(simles_file_path, 'r') as f:
    smiles_list = f.readlines()

display(smiles_list)

# 创建分子对象列表
mols = []  # 初始化一个空列表用于存储分子对象
for smiles in smiles_list:
    mol = Chem.MolFromSmiles(smiles.strip())
    mols.append(mol)

# 可视化分子对象 molsPerRow: 每行显示的分子个数, subImgSize: 每个分子的大小
img = Draw.MolsToGridImage(mols, molsPerRow=4, subImgSize=(200, 200))
display(img)

# 创建一个 SDF 文件写入对象将分子列表写入 SDF 文件
with Chem.SDWriter(sdf_file_path) as writer:
    for mol in mols:
        writer.write(mol)  

print("SDF 文件已经生成在路径：", sdf_file_path)
```

（3）sdf转换为SMILES

```python
import os
from rdkit import Chem

data_path = "data"
output_path = "output_data"
os.makedirs(output_path, exist_ok=True)
os.makedirs(data_path, exist_ok=True)
# 定义输入的 SDF 文件路径和输出的 SMI 文件路径
input_sdf_file_path = os.path.join(data_path, "20sdf.sdf")
output_smi_file_path = os.path.join(output_path, "20sdf_convert.smi")

if os.path.exists(input_sdf_file_path):
    print("文件已存在，路径为：", input_sdf_file_path)
else:
    print("文件不存在，正在从网络下载文件...")
    import requests
    file_url =  "https://raw.githubusercontent.com/ZheLi-Lab/CADD_Class_Data/main/chapter6/data/20sdf.sdf"
    response = requests.get(file_url)
    with open(input_sdf_file_path, 'wb') as f:
        f.write(response.content)
        print("文件下载成功，保存路径为：", input_sdf_file_path)

# 从 SDF 文件中读取分子信息，sanitize=False 表示不进行分子的清洗
mols = Chem.SDMolSupplier(input_sdf_file_path, sanitize=False)

# 打开输出文件以写入 SMILES 字符串
with open(output_smi_file_path, "w") as out_file:
    for mol in mols:
        smiles = Chem.MolToSmiles(mol)
        out_file.write(smiles + "\n")

print("SMILES 文件已经生成, 路径为：", output_smi_file_path)
```

（4）合并多个mol2为一个sdf

```python
import os
from rdkit import Chem

data_path = "data"
result_path = "output_data"
os.makedirs(result_path, exist_ok=True)
os.makedirs(data_path, exist_ok=True)
input_dir_path = os.path.join(data_path, "mols")
output_file_path = os.path.join(result_path, "merged_output.sdf")
if os.path.exists(input_dir_path):
    print("文件夹已存在，路径为：", input_dir_path)
else:
    import requests
    print("文件夹不存在，正在创建文件夹...")
    os.makedirs(input_dir_path, exist_ok=True)
    print("文件夹已创建，路径为：", input_dir_path)
    url_path_1 = "https://raw.githubusercontent.com/ZheLi-Lab/CADD_Class_Data/main/chapter6/data/mols/(-)-guaia-1(11)%2C11-dien-15-al.mol2"
    url_path_2 = "https://raw.githubusercontent.com/ZheLi-Lab/CADD_Class_Data/main/chapter6/data/mols/(-)-guaia-l(10)%2C11-dien-15%2C2-olide.mol2"
    url_path_3 = "https://raw.githubusercontent.com/ZheLi-Lab/CADD_Class_Data/main/chapter6/data/mols/(-)-guaia-l(10)%2C11-dien-15-ol.mol2"
    url_path_4 = "https://raw.githubusercontent.com/ZheLi-Lab/CADD_Class_Data/main/chapter6/data/mols/(-)-guaial(10)%2C11-dien-15-al.mol2"
    requests_list = [url_path_1, url_path_2, url_path_3, url_path_4]
    for url_path in requests_list:
        response = requests.get(url_path)
        file_name = url_path.split("/")[-1]
        file_path = os.path.join(input_dir_path, file_name)
        with open(file_path, "wb") as f:
            f.write(response.content)
            print(f"{file_name} 文件下载成功！路径：", file_path)

writer = Chem.SDWriter(output_file_path)    # 创建一个待写入 SDF 文件的对象
input_file_list = os.listdir(input_dir_path) # 获取输入文件夹中的所有文件名
print("input_dir_path 下的所有文件：", input_file_list)  # 打印文件名列表
for filename in input_file_list:    # 遍历文件夹中的所有文件
    if filename.endswith('.mol2'):  # 判断文件是否为 mol2 格式
        filepath = os.path.join(input_dir_path, filename)  # 拼接文件路径
        # 读取 mol2 文件中的分子信息，不进行分子清洗和去氢
        mol = Chem.MolFromMol2File(filepath, sanitize=False, removeHs=False)
        writer.write(mol)  # 将分子写入 SDF 文件
print("所有的 mol2 文件已经合并。保存路径: ", output_file_path)  # 打印提示信息
```

4.描述符生成

```python
from rdkit import Chem
from rdkit.Chem import AllChem
from rdkit.Chem import Descriptors

smiles = 'c1ccccc1'  # 定义苯分子的 SMILES
mol = Chem.MolFromSmiles(smiles)            # 从SMILES字符串中读取分子信息
mw = Descriptors.MolWt(mol)                 # 计算分子量
logP = Descriptors.MolLogP(mol)             # 计算logP值
tpsa = Descriptors.TPSA(mol)                # 计算分子的拓扑极性表面积
mol = Chem.AddHs(mol)                       # 为分子添加氢原子
AllChem.EmbedMolecule(mol)                  # 为分子生成三维坐标
conf = mol.GetConformer()                   # 获取分子的构象
volume = AllChem.ComputeMolVolume(mol, confId=0)  # 计算分子的三维结构体积

# 打印分子的属性值
print("分子量:", mw)
print("logP:", logP)
print("TPSA:", tpsa)
print("三维体积:", volume)

display(mol)    
```

5.分子指纹提取

```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem import AllChem
from rdkit.Chem import DataStructs


# 从 smiles 中读取分子
mol1 = Chem.MolFromSmiles('CC(=O)OC1=CC=CC=C1C(=O)O')
mol2 = Chem.MolFromSmiles('O=C(O)c1ccccc1O')

# 生成分子的Morgan指纹（半径为2，即直径为4的圆形指纹ECFP4）
fp1 = AllChem.GetMorganFingerprintAsBitVect(mol1, 2)
fp2 = AllChem.GetMorganFingerprintAsBitVect(mol2, 2)

# 计算指纹之间的相似性
similarity = DataStructs.TanimotoSimilarity(fp1, fp2)

# 绘制分子
img = Draw.MolsToGridImage([mol1, mol2], molsPerRow=2,
                           subImgSize=(200, 200),
                           legends=[f"Mol 1", "Mol 2"])
display(img)
print(f"两个分子之间的相似性: {similarity}")
```

6.子结构提取

```python
from rdkit import Chem
from rdkit.Chem import Draw

# 读取分子信息（阿司匹林）
mol = Chem.MolFromSmiles('CC(=O)OC1=CC=CC=C1C(=O)O')
# 定义羧基基团的子结构（C(=O)[OH]表示羧基）
substructure = Chem.MolFromSmarts('C(=O)[OH]')
# 在分子中查找羧基基团
matches = mol.GetSubstructMatches(substructure)
# 展平匹配结果，生成一个包含所有匹配原子索引的列表
print(matches)
# 定义高亮的原子索引
highlight_atoms = []
for match in matches:
    for atom in match:
        highlight_atoms.append(atom)
# 绘制分子结构并高亮匹配的子结构
img = Draw.MolToImage(mol, highlightAtoms=highlight_atoms)
display(img)
print(f"羧基基团的匹配数: {len(matches)}")
```
